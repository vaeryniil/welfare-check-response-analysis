```python
import numpy as np
import pandas as pd
import datetime as dt
```

```python
#Reading in the springfield data

#Part 1: the main data. Each sheet a different df
SPR2015 = pd.read_excel("./data/2015-2025 SPD Calls for Service.xlsx",sheet_name='2015',na_values=['      '])
SPR2016 = pd.read_excel("./data/2015-2025 SPD Calls for Service.xlsx",sheet_name='2016',na_values=['      '])
SPR2017 = pd.read_excel("./data/2015-2025 SPD Calls for Service.xlsx",sheet_name='2017',na_values=['      '])
SPR2018 = pd.read_excel("./data/2015-2025 SPD Calls for Service.xlsx",sheet_name='2018',na_values=['      '])
SPR2019 = pd.read_excel("./data/2015-2025 SPD Calls for Service.xlsx",sheet_name='2019',na_values=['      '])
SPR2020 = pd.read_excel("./data/2015-2025 SPD Calls for Service.xlsx",sheet_name='2020',na_values=['      '])
SPR2021 = pd.read_excel("./data/2015-2025 SPD Calls for Service.xlsx",sheet_name='2021',na_values=['      '])
SPR2022 = pd.read_excel("./data/2015-2025 SPD Calls for Service.xlsx",sheet_name='2022',na_values=['      '])
SPR2023 = pd.read_excel("./data/2015-2025 SPD Calls for Service.xlsx",sheet_name='2023',na_values=['      '])
SPR2024 = pd.read_excel("./data/2015-2025 SPD Calls for Service.xlsx",sheet_name='2024',na_values=['      '])
SPR2025 = pd.read_excel("./data/2015-2025 SPD Calls for Service.xlsx",sheet_name='2025',na_values=['      '])
```

```python
#Part 2: the responding units. Each sheet a different df
UNIT2015 = pd.read_excel('./data/2015-2025 SPD Responding Units.xlsx',sheet_name='2015',na_values=['      '])
UNIT2016 = pd.read_excel('./data/2015-2025 SPD Responding Units.xlsx',sheet_name='2016',na_values=['      '])
UNIT2017 = pd.read_excel('./data/2015-2025 SPD Responding Units.xlsx',sheet_name='2017',na_values=['      '])
UNIT2018 = pd.read_excel('./data/2015-2025 SPD Responding Units.xlsx',sheet_name='2018',na_values=['      '])
UNIT2019 = pd.read_excel('./data/2015-2025 SPD Responding Units.xlsx',sheet_name='2019',na_values=['      '])
UNIT2020 = pd.read_excel('./data/2015-2025 SPD Responding Units.xlsx',sheet_name='2020',na_values=['      '])
UNIT2021 = pd.read_excel('./data/2015-2025 SPD Responding Units.xlsx',sheet_name='2021',na_values=['      '])
UNIT2022 = pd.read_excel('./data/2015-2025 SPD Responding Units.xlsx',sheet_name='2022',na_values=['      '])
UNIT2023 = pd.read_excel('./data/2015-2025 SPD Responding Units.xlsx',sheet_name='2023',na_values=['      '])
UNIT2024 = pd.read_excel('./data/2015-2025 SPD Responding Units.xlsx',sheet_name='2024',na_values=['      '])
UNIT2025 = pd.read_excel('./data/2015-2025 SPD Responding Units.xlsx',sheet_name='2025',na_values=['      '])
```

```python
#Part 3: close codes. Each sheet a different df
CLOSE2015 = pd.read_excel('./data/2015-2025 SPD Calls with Close Codes.xlsx',sheet_name='2015',na_values='    ')
CLOSE2016 = pd.read_excel('./data/2015-2025 SPD Calls with Close Codes.xlsx',sheet_name='2016',na_values='    ')
CLOSE2017 = pd.read_excel('./data/2015-2025 SPD Calls with Close Codes.xlsx',sheet_name='2017',na_values='    ')
CLOSE2018 = pd.read_excel('./data/2015-2025 SPD Calls with Close Codes.xlsx',sheet_name='2018',na_values='    ')
CLOSE2019 = pd.read_excel('./data/2015-2025 SPD Calls with Close Codes.xlsx',sheet_name='2019',na_values='    ')
CLOSE2020 = pd.read_excel('./data/2015-2025 SPD Calls with Close Codes.xlsx',sheet_name='2020',na_values='    ')
CLOSE2021 = pd.read_excel('./data/2015-2025 SPD Calls with Close Codes.xlsx',sheet_name='2021',na_values='    ')
CLOSE2022 = pd.read_excel('./data/2015-2025 SPD Calls with Close Codes.xlsx',sheet_name='2022',na_values='    ')
CLOSE2023 = pd.read_excel('./data/2015-2025 SPD Calls with Close Codes.xlsx',sheet_name='2023',na_values='    ')
CLOSE2024 = pd.read_excel('./data/2015-2025 SPD Calls with Close Codes.xlsx',sheet_name='2024',na_values='    ')
CLOSE2025 = pd.read_excel('./data/2015-2025 SPD Calls with Close Codes.xlsx',sheet_name='2025',na_values='    ')
```

```python
#combining into a single SPR welfare data frame [still with no primary units or disposition]
frames = [SPR2015, SPR2016, SPR2017, SPR2018, SPR2019, SPR2020,
          SPR2021, SPR2022, SPR2023, SPR2024, SPR2025]

SPR = pd.concat(frames)
SPR
```

```python
#combining into a single SPR unit data frame
frames = [UNIT2015, UNIT2016, UNIT2017, UNIT2018, UNIT2019, UNIT2020,
          UNIT2021, UNIT2022, UNIT2023, UNIT2024, UNIT2025]
units_all_years = pd.concat(frames)
units_all_years.drop(['Unnamed: 1','call time'],axis=1,inplace=True)
units_all_years
```

```python
#combining into a single SPR closed_as data frame
frames = [CLOSE2015, CLOSE2016, CLOSE2017, CLOSE2018, CLOSE2019, CLOSE2020,
          CLOSE2021, CLOSE2022, CLOSE2023, CLOSE2024, CLOSE2025]
closed_all_years = pd.concat(frames)
closed_all_years
```

```python
#left join units
intermediate = pd.merge(SPR,units_all_years, left_on='Incident Number', right_on='inci id',how='left')
intermediate.drop('inci id',axis=1,inplace=True)
intermediate
```

```python
#left join closed codes
SPR = pd.merge(intermediate,closed_all_years, on='Incident Number',how='left')
SPR
```

```python
#Make new columns for month and year, for later grouping
SPR['month'] = SPR['Call Creation Time'].dt.month
SPR['year'] = SPR['Call Creation Time'].dt.year
```

```python
#Strip whitespace
SPR['Primary Responding Unit'] = SPR['Primary Responding Unit'].str.strip()
SPR['prime unit'] = SPR['prime unit'].str.strip()
SPR['units dispatched in order'] = SPR['units dispatched in order'].str.strip()
```

```python
#getting the count of units dispatched

#first, count number of units
unit_count = SPR['units dispatched in order'].str.count(",") + 1


#second, make new column
SPR['units dispatched'] = unit_count

#third, convert NA values to zeros
SPR['units dispatched'] = SPR['units dispatched'].fillna(0)
```

```python
#dropping rows where the primeunit is a null value and more than 0 units were dispatched
SPR = SPR[(SPR['Primary Responding Unit'].notna()) | (SPR['units dispatched'] == 0.)]
SPR
```

```python
#Adding a column saying whether cahoots, spd, or neither answered the call

SPR["Institution"] = np.where(    
    SPR["Primary Responding Unit"].str.contains("J|CAH", na=False),    
    "CAHOOTS",    
    np.where(        
        SPR["Primary Responding Unit"].isna(),        
        "Neither",        
        "SPD"    
    )
)
```
